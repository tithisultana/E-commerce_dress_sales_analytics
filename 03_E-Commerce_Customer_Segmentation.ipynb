```python
import pandas as pd
import plotly.express as px
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv("dress_sales.csv")
df["Sales"] = df["Price"] * df["Quantity"]

customer = df.groupby("Customer_ID").agg(
    Total_Spending=("Sales", "sum"),
    Total_Items=("Quantity", "sum"),
    Total_Orders=("Order_ID", "nunique")
).reset_index()

display(customer)
```

```python
features = ["Total_Spending", "Total_Items", "Total_Orders"]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(customer[features])
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
customer["Segment"] = kmeans.fit_predict(X_scaled)
display(customer.sort_values("Total_Spending", ascending=False))
```

```python
px.scatter(
    customer,
    x="Total_Spending",
    y="Total_Orders",
    size="Total_Items",
    color="Segment",
    hover_data=["Customer_ID"],
    title="Customer Segmentation"
).show()
```

```python
segment_summary = customer.groupby("Segment")[features].mean().round(2)
display(segment_summary)
```
